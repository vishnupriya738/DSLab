```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# Create TensorFlow model
# -----------------------------
model = tf.keras.Sequential([
    tf.keras.Input(shape=(3,)),          # 3 inputs
    tf.keras.layers.Dense(4, activation='relu'),
    tf.keras.layers.Dense(3, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

model.summary()


# -----------------------------
# Draw neural network
# -----------------------------
layer_sizes = [3, 4, 3, 1]

fig, ax = plt.subplots(figsize=(10, 6))

x_positions = np.arange(len(layer_sizes))

for layer_index, neurons in enumerate(layer_sizes):

    y_positions = np.linspace(
        -(neurons - 1) / 2,
        (neurons - 1) / 2,
        neurons
    )

    # Draw neurons
    for neuron_index, y in enumerate(y_positions):

        circle = plt.Circle(
            (x_positions[layer_index], y),
            0.12,
            fill=False,
            linewidth=2
        )

        ax.add_patch(circle)

        # Labels
        if layer_index == 0:
            label = f"x{neuron_index + 1}"

        elif layer_index == len(layer_sizes) - 1:
            label = "y"

        else:
            label = f"h{layer_index}{neuron_index + 1}"

        ax.text(
            x_positions[layer_index],
            y,
            label,
            ha='center',
            va='center'
        )

    # Draw connections
    if layer_index < len(layer_sizes) - 1:

        next_neurons = layer_sizes[layer_index + 1]

        next_y_positions = np.linspace(
            -(next_neurons - 1) / 2,
            (next_neurons - 1) / 2,
            next_neurons
        )

        for y1 in y_positions:
            for y2 in next_y_positions:

                ax.plot(
                    [x_positions[layer_index] + 0.12,
                     x_positions[layer_index + 1] - 0.12],
                    [y1, y2],
                    linewidth=0.8
                )


# Layer names
layer_names = [
    "Input Layer\n3 Neurons",
    "Hidden Layer 1\n4 Neurons",
    "Hidden Layer 2\n3 Neurons",
    "Output Layer\n1 Neuron"
]

for i, name in enumerate(layer_names):

    ax.text(
        x_positions[i],
        2.6,
        name,
        ha='center',
        fontsize=11,
        fontweight='bold'
    )


ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-3, 3)

ax.axis('off')

plt.title(
    "Neural Network with 3 Inputs and 2 Hidden Layers",
    fontsize=14,
    fontweight='bold'
)

plt.show()
```
